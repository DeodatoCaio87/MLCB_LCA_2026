```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

# ============================================================
# 1. CARREGAR DATASET
# ============================================================

df = pd.read_csv('dataset_moveis_100.csv')

X = df['texto']
y = df['intencao']

# ============================================================
# 2. SEPARAR TREINO E TESTE
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# ============================================================
# 3. CRIAR PIPELINE
# ============================================================

pipeline_tree = Pipeline([
    ('vectorizer', TfidfVectorizer()),
    ('classifier', DecisionTreeClassifier(random_state=42))
])

# ============================================================
# 4. TREINAR MODELO
# ============================================================

pipeline_tree.fit(X_train, y_train)

print("Decision Tree treinada com sucesso!")

# ============================================================
# 5. AVALIAÇÃO
# ============================================================

y_pred = pipeline_tree.predict(X_test)

print("\n=== MATRIZ DE CONFUSÃO - DECISION TREE ===")
print(confusion_matrix(y_test, y_pred))

print("\n=== RELATÓRIO DE CLASSIFICAÇÃO - DECISION TREE ===")
print(classification_report(y_test, y_pred))

# ============================================================
# 6. CONFIGURAÇÕES DO CHATBOT
# ============================================================

LIMIAR_CONFIANCA = 0.50

# Palavras relacionadas ao contexto do SAC
PALAVRAS_VALIDAS = [
    'comprar', 'preco', 'preço', 'cupom', 'adquirir', 'orcamento',
    'orçamento', 'sofa', 'sofá', 'mesa', 'jantar', 'guarda', 'roupa',
    'painel', 'tv', 'colchao', 'colchão',

    'montar', 'montagem', 'manual', 'duvida', 'dúvida',
    'parafuso', 'assistencia', 'assistência', 'armario', 'armário',
    'rack', 'berco', 'berço', 'estante',

    'trocar', 'troca', 'devolver', 'devolucao', 'devolução',
    'estorno', 'defeito', 'arranhada', 'avaria', 'cancelado',

    'indignado', 'queixa', 'reclamando', 'reclamação', 'reclamacao',
    'quebrado', 'horrivel', 'horrível', 'atendimento', 'sac',

    'pedido', 'entrega', 'entrega', 'rastreio', 'rastreamento',
    'transportadora', 'chega', 'envio', 'status', 'agendamento',
    'prazo', 'móvel', 'movel'
]

# ============================================================
# 7. FUNÇÃO DE CLASSIFICAÇÃO
# ============================================================

def classificar_mensagem(frase):

    frase_lower = frase.lower()

    # Verificar se existem palavras relacionadas ao domínio
    possui_contexto = any(
        palavra in frase_lower
        for palavra in PALAVRAS_VALIDAS
    )

    # Se não houver nenhuma palavra relacionada ao SAC,
    # encaminha diretamente para atendimento humano.
    if not possui_contexto:
        print("Desculpe, não entendi sua solicitação. Encaminhando você para um atendente humano...")
        return

    # Fazer previsão
    probs = pipeline_tree.predict_proba([frase])

    maior_prob = np.max(probs)

    intencao = pipeline_tree.predict([frase])[0]

    # Verificar limiar de confiança
    if maior_prob >= LIMIAR_CONFIANCA:

        print(f"Intenção identificada: {intencao}")
        print(f"Probabilidade: {maior_prob:.2%}")

    else:

        print("Desculpe, não entendi sua solicitação. Encaminhando você para um atendente humano...")


# ============================================================
# 8. TESTES MANUAIS
# ============================================================

print("\n=== INICIANDO BATERIA DE TESTES (8 INPUTS OBRIGATÓRIOS) ===")

for i in range(1, 9):

    print(f"\n[Teste {i}/8]")

    frase = input("Digite a frase do cliente: ").strip()

    classificar_mensagem(frase)
```
